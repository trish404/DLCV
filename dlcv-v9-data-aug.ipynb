```python
import shutil, os

def copy_images_and_labels(img_src, lbl_src, dst_split):
    dst_img = f"/kaggle/working/clc-xray-yolo/{dst_split}/images"
    dst_lbl = f"/kaggle/working/clc-xray-yolo/{dst_split}/labels"
    os.makedirs(dst_img, exist_ok=True)
    os.makedirs(dst_lbl, exist_ok=True)

    for file in os.listdir(img_src):
        if file.endswith((".jpg", ".png", ".jpeg")):
            shutil.copy(os.path.join(img_src, file), os.path.join(dst_img, file))

    for file in os.listdir(lbl_src):
        if file.endswith(".txt"):
            shutil.copy(os.path.join(lbl_src, file), os.path.join(dst_lbl, file))

copy_images_and_labels("/kaggle/input/train-clc/images", "/kaggle/input/train-clc/yolo_labels", "train")
copy_images_and_labels("/kaggle/input/clc-yolo-xray/valid/images", "/kaggle/input/clc-yolo-xray/valid/labels", "valid")
copy_images_and_labels("/kaggle/input/clc-yolo-xray/test/images", "/kaggle/input/clc-yolo-xray/test/labels", "test")
```

```python
import os

# Create the folder
os.makedirs("/kaggle/working/clc-xray-yolo", exist_ok=True)

# Then write the YAML
data_yaml = """
train: /kaggle/working/clc-xray-yolo/train/images
val: /kaggle/working/clc-xray-yolo/valid/images
test: /kaggle/working/clc-xray-yolo/test/images

nc: 13
names: ['gun-knife-wrench-pilers-scissors', 'Cans', 'CartonDrinks', 'GlassBottle', 'PlasticBottle', 'SprayCans', 'SwissArmyKnife', 'Tin', 'VacuumCup', 'blade', 'dagger', 'knife', 'scissors']
"""

with open("/kaggle/working/clc-xray-yolo/data.yaml", "w") as f:
    f.write(data_yaml)
```

```python
from ultralytics import YOLO

model = YOLO("yolov9c.pt")
model.train(
    data="/kaggle/working/clc-xray-yolo/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    project="/kaggle/working/yolo_results",
    name="clc-xray-yolov9c",
    exist_ok=True,
    task="detect"
)
```
